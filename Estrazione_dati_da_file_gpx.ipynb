```python
import xml.etree.ElementTree as ET


def estrai_nomi_gpx(file_input_xml, file_output_txt):
    # Parsing del file XML/GPX
    tree = ET.parse(file_input_xml)
    root = tree.getroot()

    # Estrazione di eventuali namespace presenti nel file
    namespace = ""
    if root.tag.startswith("{"):
        namespace = root.tag.split("}")[0] + "}"

    # Trova tutti i tag <wpt> (waypoints) e estrai i dati desiderati
    waypoints_data = []
    for wpt in root.iter(f"{namespace}wpt"):
        lat = wpt.get('lat')
        lon = wpt.get('lon')

        # Estrai il nome e l'elevazione, gestendo i casi in cui potrebbero non esistere
        name_elem = wpt.find(f"{namespace}name")
        name = name_elem.text.strip() if name_elem is not None and name_elem.text else ""

        ele_elem = wpt.find(f"{namespace}ele")
        ele = ele_elem.text.strip() if ele_elem is not None and ele_elem.text else ""

        waypoints_data.append(f"{name},{lat},{lon},{ele}")

    # Scrittura dei dati estratti all'interno del file di testo
    with open(file_output_txt, "w", encoding="utf-8") as f:
        for data_line in waypoints_data:
            f.write(f"{data_line}\n")

    print(
        f"Estrazione completata: {len(waypoints_data)} waypoint salvati in '{file_output_txt}'."
    )


# Utilizzo dello script:
# Sostituisci 'file_allegato.gpx' con il percorso reale del tuo file
estrai_nomi_gpx(r"C:\Users\DeRoseE\Downloads\Grignone.gpx", "elenco_nomi.txt")
```
